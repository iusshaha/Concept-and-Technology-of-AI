Problem 2 - Data Imputations:
Complete all the following Task:
• Dataset for the Task: "medical_student.csv"
1. Load the provided dataset and import in pandas DataFrame.
2. Check info of the DataFrame and identify column with missing (null) values.
3. For the column with missing values fill the values using various techniques we discussed above. Try
to explain why did you select the particular methods for particular column.
4. Check for any duplicate values present in Dataset and do necessary to manage the duplicate items.
{Hint: dataset.duplicated.sum()}

```python
#1.Load the dataset
df2 = pd.read_csv("medical_students.csv")
df2.head()
```

```python
#2.Check info
print("DATAFRAME INFO")
df2.info()

print("\nNULL VALUES COUNT")
print(df2.isnull().sum())

missing_cols = df2.columns[df2.isnull().sum() > 0]
print("\nColumns with missing values:\n", missing_cols)
```

```python
#3. Handle missing values
num_cols = df2.select_dtypes(include=['int64','float64']).columns
cat_cols = df2.select_dtypes(include=['object']).columns

print("Numerical columns:", list(num_cols))
print("Categorical columns:", list(cat_cols))

df2[num_cols] = df2[num_cols].fillna(df2[num_cols].mean())
for col in cat_cols:
    df2[col] = df2[col].fillna(df2[col].mode()[0])

print("\n----- NULL VALUES AFTER IMPUTATION -----")
print(df2.isnull().sum())
```

```python
#4. Detect duplicates
duplicate_count = df2.duplicated().sum()
print("\nNumber of duplicate rows:", duplicate_count)

df2 = df2.drop_duplicates()

print("After removing duplicates:", df2.duplicated().sum())
```

```python

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = pd.read_csv("titanic.csv")
titanic.head()
```

```python
#1
cols = ["Name","Pclass","Sex","Age","Fare","Survived"]
df = titanic[cols]

df_first = df[df["Pclass"] == 1]

print("Mean Fare:", df_first["Fare"].mean())
print("Median Fare:", df_first["Fare"].median())
print("Max Fare:", df_first["Fare"].max())
print("Min Fare:", df_first["Fare"].min())
```

• ’C’ for Cherbourg
• ’Q’ for Queenstown

Task:

Problem - 4:

Compare the mean survival rates (’Survived’) for the different groups in the ’Sex’ column. Draw a visual-
ization to show how the survival distributions vary by gender.



```python
#4
print(titanic.groupby("Sex")["Survived"].mean())

sns.barplot(x="Sex", y="Survived", data=titanic)
plt.title("Survival Rate by Gender")
plt.show()

```
